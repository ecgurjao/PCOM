# Função Densidade de Probabilidade

Considere uma variável aleatória $n$, a Função Densidade de Probabilidade $F_N(n)$ representa como a probabilidade de ocorrência de $n$ se distribui em função dos valores de dessa variável.

Como exemplo, considere que $x$ tem valores reais entre $-A$ e $A$, ou seja $n \in [-A, A]$. Caso a probabilidade que $x$ assuma qualquer dos valores no intevalo $[-A, A]$ seja a mesma, dizemos que $n$ tem distribuição uniforme nesse intervalo, representado por $n\ \tilde{}\ U[-A, A]$ e que neste caso é igual a $P_n = \frac{1}{2A}$ e a função densidade de probabilidade de $x$ está representada na figura abaixo.

![PDF_Uniforme](./Figuras/PDF_Uniforme.png)

Observe que a área do gráfico na figura acima é 1, pois represent a probabilidade total. Por óbvio, para a variável $x$ acima definida, a probabilido $P(n = 1,5A) = 0$.

Consideremos agora uma segunda variável aleatória $x$ que toma dois valores 0 ou $3A$ com igual probabilidade, nesse caso a FDP tem o formato

![PDF_Equiprovavel_Impulsos](./Figuras/PDF_Equiprovavel_Impulsos.png)


Uma terceira variável aleatória formada pela soma de $n$ com $x$, 
\[
y = x + n
\]
tem distribuição formada pela convolução das FDP de $n$ e de $x$, representada pela figura abaixo.

![PDF_Variavel_y_Uniforme](./Figuras/PDF_Variavel_y_Uniforme.png)


Da figura, observa-se que ao sortear um valor de $x$, um observador a partir do valor de $y$ consegue determinar o valor sorteado sem erro.

Em outra situação, caso a variável $x$ possa assumir os valores $0$ ou $1,5A$, dependendo do valor de $y$, o observador poderá errar ao indicar qual foi o valor de $x$.

![PDF_Variavel_y1_Uniforme](./Figuras/PDF_Variavel_y1_Uniforme.png)


## Distribuição Gaussiana

Uma variável aleatória com distribuição Gaussiana, com média $\mu$ e variância $\sigma^2$, $n\ \tilde{} N(\mu, \sigma^2)$ é dada por
$$
f_N(n) = \frac{1}{\sqrt{2\pi\sigma^2}}e^{-\frac{(n - \mu)^2}{2\sigma^2}}
$$

Na Figura abaixo está uma curva Gaussiana com média zero $\mu = 0$ e variância $\sigma = 1$.
